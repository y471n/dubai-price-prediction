```python
df = pd.read_csv(CSV)
print(f"Shape of the dataset is {df.shape}")
df.info()
```

```python
KEY_COLS = ["actual_worth", "meter_sale_price", "procedure_area"]
key_col_data = df[KEY_COLS]
print("nulls:\n", key_col_data.isna())
print("nulls:\n", key_col_data.isna().sum())
print("\nzeros:\n", (key_col_data == 0).sum())
print(key_col_data.describe().T[["count", "mean", "50%", "max"]])
```

```python
ok = df[df["trans_group_en"] == "Sales"][KEY_COLS].dropna()
calc = ok["actual_worth"] / ok["procedure_area"]
match = ((ok["meter_sale_price"] - calc).abs() / calc < 0.01).mean()
print(f"Sales rows: {len(ok):,}")
print(f"meter_sale_price ≈ actual_worth/procedure_area within 1%: {match:.1%}")
print(ok.describe().T[["mean", "50%", "max"]])
```

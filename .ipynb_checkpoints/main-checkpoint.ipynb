## Filtering observations by Property Classifications

`property_usage_en` column

We need to make a decision to use which all properties from `property_usage_en`.
eg: If we want to just use Residential properties, then we need to filter out the usage types of following:
1. Residential
2. Residential / Commercial
3. Industrial / Commercial / Residential
4. Multi-Use

_Multi-Use might be removed seeing it's values_

```python
allowed_usages = ['Residential', 'Residential / Commercial', 'Industrial / Commercial / Residential', 'Multi-Use']
res_df = df2[df2['property_usage_en'].isin(allowed_usages)]
```

# Target Encoding

```python
from sklearn.preprocessing import TargetEncoder


# target encode:  building_name_en project_name_en

target_encoder = TargetEncoder()

```

```python
target_encoding_columns = ['building_name_en', 'project_name_en']
target_encoding_df = X_train[target_encoding_columns]
target = y_train
target_transform = target_encoder.fit_transform(target_encoding_df, target)
```

```python
target_transform.shape
```

```python
target_transform[:, 0]
```

```python
X_train[['building_name_en_encoded', 'project_name_en_encoded']] = target_transform
```

```python
X_train.head()
```

```python
test_target_transform = target_encoder.transform(X_test[target_encoding_columns])
test_target_transform
```

```python
X_test[['building_name_en_encoded', 'project_name_en_encoded']] = test_target_transform
```

```python
X_train = X_train.drop(columns=target_encoding_columns)
X_test = X_test.drop(columns=target_encoding_columns)
```

## Filtering observations by Property Classifications

`property_usage_en` column

We need to make a decision to use which all properties from `property_usage_en`.
eg: If we want to just use Residential properties, then we need to filter out the usage types of following:
1. Residential
2. Residential / Commercial
3. Industrial / Commercial / Residential
4. Multi-Use

_Multi-Use might be removed seeing it's values_

```python
allowed_usages = ['Residential', 'Residential / Commercial', 'Industrial / Commercial / Residential', 'Multi-Use']
res_df = df2[df2['property_usage_en'].isin(allowed_usages)]
```

```python
model = xgb.XGBRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.80,
    n_jobs=-1,
    tree_method='hist',
    early_stopping_rounds=50,
    objective='reg:squarederror',
    enable_categorical=True
)

# Fit the model
model.fit(
    X_train, y_train,
    eval_set=[(X_test, y_test)],
    verbose=100
)
```

```python
from datetime import datetime, date

def get_date_features(date_str=None):
    """
    Returns:
    - days difference from 2003-06-02
    - year
    - month
    - quarter
    """

    base_date = date(2003, 6, 2)

    # Use today's date if no date string provided
    if not date_str:
        input_date = date.today()
    else:
        input_date = datetime.strptime(date_str, "%d-%m-%Y").date()

    days_diff = (input_date - base_date).days
    year = input_date.year
    month = input_date.month
    quarter = (month - 1) // 3 + 1

    return days_diff, year, month, quarter


# Examples
print(get_date_features("15-10-2020"))
print(get_date_features())        # uses today's date
print(get_date_features(None))    # also uses today's date
```

```python
import pickle
import numpy as np
import pandas as pd

def predict_price(trans_group_en, procedure_name_en, reg_type_en, building_name_en,
                  procedure_area_en, area_name_en, rooms_num,
                  project_name_en=None, master_project_en=None, nearest_landmark_en=None,
                  nearest_metro_en=None, nearest_mall_en=None, has_parking=0,
                  transaction_date=None):
    
    active_model = globals().get('model')
    if active_model is None:
        with open('dubai_flats_prices.pickle', 'rb') as f:
            active_model = pickle.load(f)

    days_since_reference_date, year, month, quarter = get_date_features(transaction_date)

    if project_name_en is None:
        project_name_en = building_name_en
    if master_project_en is None:
        master_project_en = project_name_en
    if nearest_landmark_en is None:
        nearest_landmark_en = '__MISSING__'
    if nearest_metro_en is None:
        nearest_metro_en = '__MISSING__'
    if nearest_mall_en is None:
        nearest_mall_en = '__MISSING__'

    row = {
        'trans_group_en': trans_group_en,
        'procedure_name_en': procedure_name_en,
        'reg_type_en': reg_type_en,
        'building_name_en': building_name_en,
        'project_name_en': project_name_en,
        'master_project_en': master_project_en,
        'nearest_landmark_en': nearest_landmark_en,
        'nearest_metro_en': nearest_metro_en,
        'nearest_mall_en': nearest_mall_en,
        'has_parking': int(has_parking),
        'procedure_area': float(procedure_area_en),
        'area_name_en': area_name_en,
        'rooms_num': float(rooms_num),
        'days_since_reference_date': int(days_since_reference_date),
        'year': int(year),
        'month': int(month),
        'quarter': int(quarter),
    }

    x = pd.DataFrame([row])

    if hasattr(active_model, 'feature_names_in_'):
        feature_order = list(active_model.feature_names_in_)
    elif 'X_train' in globals():
        feature_order = list(X_train.columns)
    else:
        feature_order = list(x.columns)

    for col in feature_order:
        if col not in x.columns:
            x[col] = np.nan
    x = x[feature_order]

    categorical_cols = [
        'trans_group_en', 'procedure_name_en', 'reg_type_en', 'building_name_en',
        'project_name_en', 'master_project_en', 'nearest_landmark_en',
        'nearest_metro_en', 'nearest_mall_en', 'area_name_en'
    ]
    for col in categorical_cols:
        if col in x.columns:
            x[col] = x[col].astype('category')

    log_price = float(active_model.predict(x)[0])
    price = np.exp(log_price)
    return price
```

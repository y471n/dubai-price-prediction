```python
from datetime import datetime, date

def get_date_features(date_str=None):
    """
    Returns:
    - days difference from 2023-06-02
    - year
    - month
    - quarter
    """
    base_date = date(2023, 6, 2)

    # Use today's date if no date string provided
    if not date_str:
        input_date = date.today()
    else:
        input_date = datetime.strptime(date_str, "%d-%m-%Y").date()

    days_diff = (input_date - base_date).days
    year = input_date.year
    month = input_date.month
    quarter = (month - 1) // 3 + 1

    return days_diff, year, month, quarter


# Examples
print(get_date_features("15-10-2020"))
print(get_date_features())        # uses today's date
print(get_date_features(None))    # also uses today's date
```

```python
# Use on API side
import json
import joblib
import xgboost as xgb
import numpy as np
import pandas as pd

def load_artifacts():
    global model, feature_order, maps, flats_df

    # Load model
    model = xgb.XGBRegressor()
    model.load_model("artifacts2/xgb_model.json")

    # Load feature order
    with open("artifacts2/features.json") as f:
        feature_order = json.load(f)
    
    # Load maps
    maps = {
        "building_master_project_map": joblib.load("artifacts2/building_master_project_map.pkl"),
        "building_ppm_map": joblib.load("artifacts2/building_ppm_map.pkl"),
        "building_ppm_6m_map": joblib.load("artifacts2/building_ppm_6m_map.pkl"),
        "master_project_ppm_map": joblib.load("artifacts2/master_project_ppm_map.pkl"),
        "master_project_ppm_3m_map": joblib.load("artifacts2/master_project_ppm_3m_map.pkl"),
    }

    # Flats df
    flats_df = joblib.load("artifacts2/flats_df.pkl")


def preprocess_input(data, maps):
    building_name_en = data["building_name_en"]
    procedure_area = data["procedure_area"]
    rooms_num = data["rooms_num"]

    building_master_project_map = maps["building_master_project_map"]
    building_ppm_map = maps["building_ppm_map"]
    building_ppm_6m_map = maps["building_ppm_6m_map"]
    master_project_ppm_map = maps["master_project_ppm_map"]

    # Validate
    if building_name_en not in building_master_project_map:
        return None

    master_project_en = building_master_project_map[building_name_en]

    # Base values
    building_ppm = building_ppm_map.get(building_name_en, np.nan)
    master_project_ppm = master_project_ppm_map.get(master_project_en, np.nan)
    building_ppm_6m = building_ppm_6m_map.get(building_name_en, np.nan)

    # Fallback
    if pd.isna(building_ppm):
        building_ppm = master_project_ppm

    # Derived features
    building_premium = (
        building_ppm / master_project_ppm
        if pd.notna(building_ppm) and pd.notna(master_project_ppm) and master_project_ppm > 0
        else np.nan
    )

    building_price_est = (
        building_ppm * procedure_area
        if pd.notna(building_ppm)
        else np.nan
    )

    master_project_price_est = (
        master_project_ppm * procedure_area
        if pd.notna(master_project_ppm)
        else np.nan
    )

    building_discount = (
        building_ppm_6m - master_project_ppm
        if pd.notna(building_ppm_6m) and pd.notna(master_project_ppm)
        else np.nan
    )

    # Log features
    log_area = np.log1p(procedure_area)

    log_building_price_est = (
        np.log1p(building_price_est)
        if pd.notna(building_price_est)
        else np.nan
    )

    log_master_project_price_est = (
        np.log1p(master_project_price_est)
        if pd.notna(master_project_price_est)
        else np.nan
    )

    log_building_premium = (
        np.log1p(building_premium)
        if pd.notna(building_premium)
        else np.nan
    )

    # Date features
    days_since_reference_date, year, month, quarter = get_date_features()

    # Final row
    row = {
        "building_name_en": building_name_en,
        "master_project_en": master_project_en,
        "rooms_num": float(rooms_num),

        "log_area": log_area,
        "building_ppm": building_ppm,
        "building_ppm_6m": building_ppm_6m,

        "building_premium": building_premium,
        "log_building_premium": log_building_premium,

        "log_building_price_est": log_building_price_est,
        "log_master_project_price_est": log_master_project_price_est,

        "building_discount": building_discount,

        "days_since_reference_date": days_since_reference_date,
        "year": int(year),
        "month": int(month),
        "quarter": int(quarter),
    }

    return pd.DataFrame([row])

def comparable_transactions(
    building_name_en,
    procedure_area,
    rooms_num,
    days_since_reference_date,
    df  # pass your loaded flats_df here
):
    # 1. Filter same building
    df_building = df[df["building_name_en"] == building_name_en].copy()

    if df_building.empty:
        return []

    # 2. Filter similar rooms (+/- 1)
    df_rooms = df_building[
        df_building["rooms_num"].between(rooms_num - 1, rooms_num + 1)
    ].copy()

    if df_rooms.empty:
        return []

    # 3. Recover actual area from log
    df_rooms["procedure_area"] = np.expm1(df_rooms["log_area"])

    # 4. Filter similar size (70% – 120%)
    df_size = df_rooms[
        df_rooms["procedure_area"].between(
            0.7 * procedure_area,
            1.2 * procedure_area
        )
    ].copy()

    if df_size.empty:
        return []

    # 5. Filter recent transactions
    df_recent = df_size[
        df_size["days_since_reference_date"].between(
            days_since_reference_date - 200,
            days_since_reference_date + 50
        )
    ].copy()

    if df_recent.empty:
        return []

    # 6. Recover price from log
    df_recent["price"] = np.round(np.expm1(df_recent["actual_worth_log"]))

    # 7. Sort and limit
    df_recent = df_recent.sort_values(
        "days_since_reference_date",
        ascending=False
    )

    result = df_recent[
        ["building_name_en", "area_name_en", "rooms_num", "price"]
    ].head(5)

    return result.to_dict(orient="records")

def predict(data):
    x = preprocess_input(data, maps)

    if x is None:
        return 0

    # 🔥 Enforce feature contract
    for col in feature_order:
        if col not in x.columns:
            print(f"{col} not found")
            x[col] = np.nan

    x = x[feature_order]
    assert set(feature_order) == set(x.columns), "Feature mismatch!"

    # ensure categorical dtype
    for col in x.select_dtypes(include="str").columns:
        x[col] = x[col].astype("category")

    # Predict
    log_price = model.predict(x)[0]
    price = float(np.expm1(log_price))


    return price, x


### PREDICTION
test_unit = {
    'building_name_en': 'Socio Tower 2',
    'procedure_area': 45,
    'rooms_num': 1
}
load_artifacts()
price, parsed_unit_df = predict(test_unit)
print(price)
### PREDICTION


### COMPARABLE TRANSACTIONS

comparables = comparable_transactions(
        test_unit['building_name_en'],
        test_unit['procedure_area'],
        test_unit['rooms_num'],
        parsed_unit_df['days_since_reference_date'].iloc[0],
        flats_df
    )
print(comparables)
### COMPARABLE TRANSACTIONS
```

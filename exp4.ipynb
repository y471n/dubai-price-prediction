```python
from datetime import datetime, date

def get_date_features(date_str=None):
    """
    Returns:
    - days difference from 2023-06-02
    - year
    - month
    - quarter
    """
    base_date = date(2023, 6, 2)

    # Use today's date if no date string provided
    if not date_str:
        input_date = date.today()
    else:
        input_date = datetime.strptime(date_str, "%d-%m-%Y").date()

    days_diff = (input_date - base_date).days
    year = input_date.year
    month = input_date.month
    quarter = (month - 1) // 3 + 1

    return days_diff, year, month, quarter


# Examples
print(get_date_features("15-10-2020"))
print(get_date_features())        # uses today's date
print(get_date_features(None))    # also uses today's date
```

```python
def _safe_mode(series):
    mode = series.mode(dropna=True)
    if not mode.empty:
        return mode.iloc[0]
    non_null = series.dropna()
    return non_null.iloc[0] if not non_null.empty else None

building_area_map = (
    final_df.groupby('building_name_en')['area_name_en']
    .apply(_safe_mode)
    .dropna()
    .to_dict()
 )
```

```python
def comparable_transactions(building_name_en,
                  procedure_area, rooms_num,
                days_since_reference_date):
    same_building_txns = final_df[final_df['building_name_en'] == building_name_en]
    similar_rooms_txns = same_building_txns[same_building_txns['rooms_num'].between(rooms_num - 1, rooms_num + 1)]
    similar_size_txns = similar_rooms_txns[(similar_rooms_txns['procedure_area'] >= 0.7 * procedure_area) & (similar_rooms_txns['procedure_area'] <= 1.2 * procedure_area)]
    recent_txns = similar_size_txns[similar_size_txns['days_since_reference_date'].between(days_since_reference_date - 200, days_since_reference_date + 50)]
    recent_txns['price'] = np.expm1(recent_txns['actual_worth_log'])
    return recent_txns.sort_values('days_since_reference_date', ascending=False)[:5]
```

```python
import pickle
import numpy as np
import pandas as pd

DEFAULT_VALUATION_DATE = '01-01-2025'

def _lookup_engineered_features(building_name_en, area_name_en, days_since_reference_date):
    b_hist = final_df[final_df['building_name_en'] == building_name_en]
    a_hist = final_df[final_df['area_name_en'] == area_name_en]

    b_recent = b_hist[b_hist['days_since_reference_date'].between(days_since_reference_date - 180, days_since_reference_date)]
    a_recent = a_hist[a_hist['days_since_reference_date'].between(days_since_reference_date - 90, days_since_reference_date)]

    building_ppm_6m = b_recent['building_ppm_6m'].median()
    if pd.isna(building_ppm_6m):
        building_ppm_6m = b_hist['building_ppm_6m'].median()
    if pd.isna(building_ppm_6m):
        building_ppm_6m = final_df['building_ppm_6m'].median()

    area_ppm_3m = a_recent['area_ppm_3m'].median()
    if pd.isna(area_ppm_3m):
        area_ppm_3m = a_hist['area_ppm_3m'].median()
    if pd.isna(area_ppm_3m):
        area_ppm_3m = final_df['area_ppm_3m'].median()

    area_ppm = a_hist['area_ppm'].median()
    if pd.isna(area_ppm):
        area_ppm = final_df['area_ppm'].median()

    return float(building_ppm_6m), float(area_ppm_3m), float(area_ppm)

def _compute_building_premium(building_ppm_6m, area_ppm):
    if pd.notna(area_ppm) and area_ppm != 0:
        return float(building_ppm_6m / area_ppm)

    premium_series = (final_df['building_ppm_6m'] / final_df['area_ppm']).replace([np.inf, -np.inf], np.nan)
    premium_median = premium_series.median()
    return float(premium_median) if pd.notna(premium_median) else np.nan

def _default_date_features():
    return get_date_features(DEFAULT_VALUATION_DATE)

def predict_price(building_name_en, procedure_area, rooms_num, transaction_date=None):
    area_name_en = building_area_map.get(building_name_en)
    if area_name_en is None:
        raise ValueError(f"Unknown building_name_en: {building_name_en}")

    active_model = globals().get('model')
    if active_model is None:
        with open('dubai_flats_prices.pickle', 'rb') as f:
            active_model = pickle.load(f)

    if transaction_date is None:
        days_since_reference_date, year, month, quarter = _default_date_features()
        used_default_valuation_date = True
        effective_transaction_date = DEFAULT_VALUATION_DATE
    else:
        days_since_reference_date, year, month, quarter = get_date_features(transaction_date)
        used_default_valuation_date = False
        effective_transaction_date = transaction_date

    building_ppm_6m, area_ppm_3m, area_ppm = _lookup_engineered_features(
        building_name_en, area_name_en, days_since_reference_date
    )
    building_premium = _compute_building_premium(building_ppm_6m, area_ppm)

    row = {
        'building_name_en': building_name_en,
        'area_name_en': area_name_en,
        'rooms_num': float(rooms_num),
        'procedure_area': float(procedure_area),
        'building_premium': float(building_premium) if pd.notna(building_premium) else np.nan,
        'building_ppm_6m': float(building_ppm_6m),
        'area_ppm_3m': float(area_ppm_3m),
        'area_ppm': float(area_ppm),
        'days_since_reference_date': int(days_since_reference_date),
        'year': int(year),
        'month': int(month),
        'quarter': int(quarter),
    }

    x = pd.DataFrame([row])

    if hasattr(active_model, 'feature_names_in_'):
        feature_order = [str(c) for c in active_model.feature_names_in_]
    elif 'X_train' in globals():
        feature_order = list(X_train.columns)
    else:
        feature_order = list(x.columns)

    for col in feature_order:
        if col not in x.columns:
            x[col] = np.nan
    x = x[feature_order]

    categorical_cols = x.select_dtypes(include=['object', 'string']).columns
    for col in categorical_cols:
        x[col] = x[col].astype('category')

    raw_model_output = float(np.expm1(active_model.predict(x)[0]))

    related_transactions = comparable_transactions(
        building_name_en, procedure_area, rooms_num, days_since_reference_date
    )

    use_ppm_interpretation = False
    predicted_total = raw_model_output

    if not related_transactions.empty:
        comparable_median = float(related_transactions['price'].median())
        as_total = raw_model_output
        as_ppm_total = raw_model_output * float(procedure_area)
        if abs(as_ppm_total - comparable_median) < abs(as_total - comparable_median):
            predicted_total = as_ppm_total
            use_ppm_interpretation = True

    predict_price.last_debug = {
        'raw_model_output': raw_model_output,
        'used_ppm_interpretation': use_ppm_interpretation,
        'used_default_valuation_date': used_default_valuation_date,
        'effective_transaction_date': effective_transaction_date,
        'days_since_reference_date': int(days_since_reference_date),
        'year': int(year),
        'month': int(month),
        'quarter': int(quarter),
        'area_name_en': area_name_en,
        'building_premium': building_premium,
        'building_ppm_6m': building_ppm_6m,
        'area_ppm_3m': area_ppm_3m,
        'area_ppm': area_ppm,
    }

    return predicted_total, related_transactions
```
